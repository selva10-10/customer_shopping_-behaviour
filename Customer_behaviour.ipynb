```python
df["Review Rating"]=df.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower().str.replace(" ","_")
```

```python
df.columns
```

```python
df.columns=df.columns.str.replace("purchase_amount_(usd)","purchase_amount")
```

```python
df["age_group"]=pd.cut(df["age"],bins=[0,18,35,60,100],labels=["child","youth","adult","senior"])
```

```python
frequency_of_mapping={'Fortnightly':14, 'Weekly':7, 'Annually':365, 'Quarterly':91, 'Bi-Weekly':14,
       'Monthly':30, 'Every 3 Months':90}
df["purchase_of_frequency"]=df["frequency_of_purchases"].map(frequency_of_mapping)
```
